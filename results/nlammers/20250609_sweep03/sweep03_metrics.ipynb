### Preliminary conclusions

1. Swin-Tiny and Efficient-Net V4 are the clear winners. Efficient net is indeed "efficient", achieving comparable recon scores at only about 0.75 the VRAM footprint of Swin-Tiny
2. It seems clear that bigger is better, more or less. Could be worth trying to resolve NaN issues with larger nets in case this drives substantial additional gains.
3. Training with PIPs-only prior to GAN seems to improve performance; however this is a bit suspect since a different pips net is used to score images
4. I clearly need to down-weight the GAN term. This can be seen both by the fact that the lower gan weight (0.05 vs 0.25) consistently performs better. We also generally see that grad_D is signficantly larger than grad_G, which seems to run counter to recommendation.
