For all of these trends, it behooves us to dig into the underlying cell types and tissues to figure out what is going on.

```python
import pandas as pd
import numpy as np
from tqdm import tqdm
from itertools import product

# n_emb_thresh = 8
group_var = "cell_type"
n_boot = 100   # Number of bootstrap iterations
ci_alpha = 0.05

# Get unique group values before bootstrapping
tissue_values = col_df[group_var].astype(str).unique()
embryo_values = col_df['embryo_ID'].unique()
temp_values = col_df['temp'].unique()
timepoint_values = col_df['timepoint'].unique()
standard_cell_types = np.unique(tissue_values)
temp_vec = np.asarray(sorted(pd.Series(temp_values).dropna().unique())).astype(float)

np.random.seed(371)

# ---- BOOTSTRAP LOOP ---- #
boot_corrs_by_temp = {temp: [] for temp in temp_vec}
min_group_n = 3         # Minimum number of rows in each group to aggregate
min_embryos_per_temp = 5  # Minimum number of embryos at each temp to calculate correlations

# Filter col_df to keep only groups with enough data for aggregation
group_counts = col_df.groupby(['embryo_ID', group_var, 'temp', 'timepoint']).size().reset_index(name='n')
keep_mask = group_counts['n'] >= min_group_n
good_groups = group_counts.loc[keep_mask, ['embryo_ID', group_var, 'temp', 'timepoint']]

# Merge to keep only "good" groups in col_df
col_df_filtered = pd.merge(col_df, good_groups, on=['embryo_ID', group_var, 'temp', 'timepoint'], how='inner')

# Proceed as before, replacing col_df with col_df_filtered in the bootstrap
for boot in tqdm(range(n_boot), desc="Bootstrapping group means"):
    boot_sample_means = (
        col_df_filtered
        .groupby(['embryo_ID', group_var, 'temp', 'timepoint'], group_keys=False)
        .apply(lambda g: g.sample(n=len(g), replace=True)['mean_nn_time'].mean())
        .reset_index(name='mean_nn_time')
    )
    for temp in temp_vec:
        temp_df = boot_sample_means[boot_sample_means["temp"] == temp]
        pivot_df = temp_df.pivot(index='embryo_ID', columns=group_var, values='mean_nn_time')
        pivot_df = pivot_df.reindex(columns=standard_cell_types)
        if len(pivot_df) >= min_embryos_per_temp:
            corr = pivot_df.corr(method='pearson')
            boot_corrs_by_temp[temp].append(corr.to_numpy())
        else:
            boot_corrs_by_temp[temp].append(np.full((len(standard_cell_types), len(standard_cell_types)), np.nan))


# ---- POINT ESTIMATE TABLE ---- #
emb_tissue_stage_df = (
    col_df_filtered
    .groupby(['embryo_ID', group_var, 'temp', 'timepoint'])
    ['mean_nn_time']
    .agg(['mean', 'size'])
    .reset_index()
    .rename(columns={"mean":"mean_nn_time", "size":"n_cells"})
)
correlations = {}
for temp in temp_vec:
    temp_df = emb_tissue_stage_df[emb_tissue_stage_df["temp"] == temp]
    pivot_df = temp_df.pivot(index='embryo_ID', columns=group_var, values='mean_nn_time')
    pivot_df = pivot_df.reindex(columns=standard_cell_types)
    correlations[temp] = pivot_df.corr(method='pearson')
```

```python
output_dir = "./corr_ci_results"  # Change to your output directory
os.makedirs(output_dir, exist_ok=True)

for temp in temp_vec:
    # Stack bootstrap results into array (n_boot, n_cell_types, n_cell_types)
    boot_arr = np.stack(boot_corrs_by_temp[temp])  # shape: (n_boot, n_types, n_types)
    
    # Compute statistics across bootstrap dimension
    ci_lower = np.nanpercentile(boot_arr, 2.5, axis=0)
    ci_upper = np.nanpercentile(boot_arr, 97.5, axis=0)
    median_corr = np.nanmedian(boot_arr, axis=0)
    
    # Get point estimate (from full data)
    corr_df = correlations[temp]  # DataFrame shape: (n_types, n_types)
    
    # Convert to DataFrame for saving, add index/col names
    corr_df_out = pd.DataFrame(corr_df, index=standard_cell_types, columns=standard_cell_types)
    ci_lower_df = pd.DataFrame(ci_lower, index=standard_cell_types, columns=standard_cell_types)
    ci_upper_df = pd.DataFrame(ci_upper, index=standard_cell_types, columns=standard_cell_types)
    median_corr_df = pd.DataFrame(median_corr, index=standard_cell_types, columns=standard_cell_types)
    
    # Save to CSV
    temp_str = f"temp_{temp}".replace(".", "_")
    corr_df_out.to_csv(f"{fig_folder}/cell_corr_matrix_{temp_str}.csv")
    ci_lower_df.to_csv(f"{fig_folder}/cell_corr_ci_lower_{temp_str}.csv")
    ci_upper_df.to_csv(f"{fig_folder}/cell_corr_ci_upper_{temp_str}.csv")
    median_corr_df.to_csv(f"{fig_folder}/cell_corr_median_{temp_str}.csv")
```
